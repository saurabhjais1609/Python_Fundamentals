```python
# Dictionary
# Definition: A dictionary is nothing but a combination of key value pair.
# Classification: It is classified as a mutable datatype.
# How to declare the dictionary --> {}
```

```python
# Alien Game
alien = {'color':'green', 'points':'5', 'level':'stage-1'}
print(alien)
```

```python
print(alien['color'])
```

```python
# Instagram User Account
useraccount = {'username':'imsaurabhjais_16', 'firstname':'Saurabh', 'lastname':'Jaiswal', 'dob':'16-09-1999', 'password':'12345'}
print(useraccount)
```

```python
#To access username
print(useraccount['username'])
```

```python
#To modify password
useraccount['password']='54321'
print(useraccount)
```

```python
#To delete lastname
del useraccount['lastname']
print(useraccount)
```

```python
# Iterating through a dictionary
useraccount = {'username':'imsaurabhjais_16', 'firstname':'Saurabh', 'lastname':'Jaiswal', 'dob':'16-09-1999', 'password':'12345'}
```

```python
#To iterate key-value pair
for key, value in useraccount.items():
    print(f"{key}:{value}")
```

```python
#To iterate key
for key in useraccount:
    print(key)
```

```python
for key in useraccount.keys():
    print(key)
```

```python
#To iterate value
for value in useraccount.values():
    print(value)
```

```python
# Nested Dictionaries
d = {
    "student": {
        "name": "Sam",
        "age": 20
    }
}

print(d["student"]["name"])
```
