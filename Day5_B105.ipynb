```python
# Organising the element in the list
# Organise the element in the list in Alphabetical Order --> A-Z
    # 1. Temporary Approach ----> sorted() 
        # sorted() returns a new list containing its elements in sorted order. The original iterable remains unchanged.
    # 2. Permanent Approach ----> sort()
        # sort() changes the original list and return none.
```

```python
cars = ['bmw', 'suzuki', 'toyota', 'audi', 'honda', 'benz', 'range rover']
print(cars)
```

```python
# sorted()
print(sorted(cars))
print(cars)
```

```python
cars.sort()
print(cars)
```

```python
# Interview Question: Difference between sorted() and sort() in the list datatype.
```

```python
# Organise the element in the list in Reverse Order --> Z-A
# sort(reverse=True), reverse()
```

```python
cars.sort(reverse=True)
print(cars)
```

```python
cars.sort()
print(cars)
cars.reverse()
print(cars)
```

```python
# To know the number of element
len(cars)
```

```python
# Slicing
# List slicing is a technique used to extract a portion of a list by specifying a range of indices. 
# It returns a new list containing the selected elements while leaving the original list unchanged.
# General Syntax: list_name[start_value : end_value : step_count]
# Note: End/Last value is always exclusive to include increment the index by +1.
```

```python
fruits = ["Apple", "Banana", "Orange", "Mango", "Grapes", "Kiwi"]
print(fruits)
```

```python
# To get Apple and Banana
print(fruits[0:2])
```

```python
# To get Orange and Mango
print(fruits[2:4])
```

```python
# To get Grapes and Kiwi
print(fruits[4:6])
```

```python
# To get Apple, Orange and Kiwi
print(fruits[0:6:2])
```
