```python
# Tuple
# Definition: An immutable list is called as a tuple. (Data is fixed --> we cannot edit/modify the data)
# Classification: It is an immutable datatype. (fixed in nature)
# How to declare the tuple datatype --> ()
```

```python
students = ('Priya', 'Vishnu', 'Karthik', 'Mahesh', 'Aman', 'Saima', 'Sana')
print(students)
```

```python
print(students[1])
```

```python
type(students)
```

```python
tup = ()
print(tup)
type(tup)
```

```python
li = [1, 2, 4, 5, 6]
print(tuple(li))
```

```python
name = tuple('Saurabh')
print(name)
```
