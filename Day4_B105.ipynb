```python
# List
# Definition: A list is a collection of items declared in a particular order.
# Classification: It is classified as mutable datatype.
# How to declare the list 
    # 1. Using square brackets []
        # a = [1, 2, 3]
    # 2. Using list() constructor
        # a = list((1,2,3))
```

```python
students = ['Radhika', 'Anjali', 'Renuka', 'Monika', 'Meera', 'Sonali']
print(students)
```

```python
type(students)
```

```python
# Indexing
# 0,1,2,3,...
```

```python
#Accessing Element
print(students[2])
```

```python
# How to add/modify/delete element in the list 
# Adding Element 
# append(), insert(), extends()
```

```python
students.append('Shiwanjali')
print(students)
```

```python
students.insert(0, 'Anamika')
print(students)
```

```python
# Interview Question: Difference between append and insert method in a list datatype.
```

```python
# Updating/Modifying ELement
students[7] = 'Monali'
print(students)
```

```python
# Delete/Removing Element
# remove(), pop(), del statement, clear()
```

```python
students.pop()
print(students)
```

```python
students.remove('Anamika')
print(students)
```

```python
del students[4]
print(students)
```

```python
students.clear()
print(students)
```
