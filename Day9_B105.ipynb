```python
# Introduction to Conditional Statements
# Conditional statements are used to control the flow of execution in a program based on specific conditions. 
```

```python
# Zen of python:
import this
```

```python
# If Statement
# If statement is used to execute a block of code only when a specified condition evaluates to True.
age = 20
if age >= 18:
    print("Eligible to vote.")
```

```python
# If Else Statement
# If Else statement is used to execute one block of code when the condition is True and another block when the condition is False.
age = 10
if age <= 12:
    print("Travel for free.")
else:
    print("Pay for ticket.")
```

```python
# If-elif-else Statement
# elif statement is used to check multiple conditions in a program. 
# It executes a block of code when its condition evaluates to True after previous conditions evaluate to False.
age = 25
if age <= 12:
    print("Child.")
elif age <= 19:
    print("Teenager.")
elif age <= 35:
    print("Young adult.")
else:
    print("Adult.")
```

```python
# Example
# Ticket Counter
# Adult: Age>10, Rs 20
# Kid: Age<10 Rs 10
# Senior Citizen: Age>=60, Rs 5

person_age = 30
if person_age<10:
    print("You are a kid, please pay Rs 10 as a ticket price")
elif person_age>=60:
    print("You are an senior citizen, please pay Rs 5 as a ticket price")
else: 
    print("You are an adult, please pay Rs 20 as a ticket price")
```

```python
# Nested if else Statement
# Nested if else statement is an if-else statement placed inside another if or else block. It is used to check conditions within another condition.
age = 70
is_member = True

if age >= 60:
    if is_member:
        print("30% senior discount!")
    else:
        print("20% senior discount.")
else:
    print("Not eligible for a senior discount.")
```

```python
# Conditional Expression
# Conditional Expression (Ternary Operator) is a short way to write an if-else statement in a single line. 
# It is used when choosing between two values based on a condition.
age = 20
s = "Adult" if age >= 18 else "Minor"
print(s)
```
