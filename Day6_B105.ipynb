```python
# Loops
# Loops are fundamental constructs in programming that allow you to execute a block of code repeatedly.
```

```python
# for loop
# Python for loops are used to iterate over sequences such as lists, tuples, strings and ranges.

# General Syntax
# for variable in sequence:
    # code block
```

```python
name = "Saurabh"
for char in name:
    print(char)
```

```python
students = ['Saurabh', 'Shiwanjali', 'Anjali', 'Sonali', 'Renuka', 'Monika', 'Meera']
for student_name in students:
    print(student_name)
```

```python
for student_name in students:
    print(f"Keep up the good work, {student_name}")
```

```python
for student_name in students:
print(student_name)
```
